```python
#step3 display datatypes
print(df.info())
```

Part 2: Data Visualization 

```python
#step8 histogram
# Histogram of numeric columns
numeric_cols = ['Year', 'Engine_Size', 'Mileage', 'Horsepower', 'Torque', 'Owners', 'Fuel_Efficiency', 'Selling_Price']

df[numeric_cols].hist(figsize=(14, 10), bins=15, edgecolor='black')
plt.tight_layout()
plt.show()

plt.figure(figsize=(6, 4))
sns.histplot(df['Fuel_Efficiency'], bins=20, kde=True)
plt.title('Distribution of Fuel Efficiency')
plt.show()

plt.figure(figsize=(6, 4))
sns.histplot(df['Selling_Price'], bins=20, kde=True)
plt.title('Distribution of Selling Price')
plt.show()
```

Part 3: Data Cleaning

```python
#step10 missing value handling
# fill with median
numeric_missing = ['Engine_Size', 'Horsepower', 'Torque', 'Accident_History', 'Fuel_Efficiency']
for col in numeric_missing:
    df[col] = df[col].fillna(df[col].median())

# fill with mode 
categorical_missing = ['Transmission', 'Service_History', 'Color', 'Location']
for col in categorical_missing:
    df[col] = df[col].fillna(df[col].mode()[0])
    
print(df.isnull().sum())
```

```python
#step11 duplicate removal
print('Num of duplicated',df.duplicated().sum())

#dont have any duplicates so no need to remove
```

```python
#step12 incorrect data correction
    #Check for negative values
print("Negative Year:", (df['Year'] < 0).sum())
print("Negative Mileage:", (df['Mileage'] < 0).sum())
print("Negative Engine_Size:", (df['Engine_Size'] < 0).sum())
print("Negative Horsepower:", (df['Horsepower'] < 0).sum())
print("Negative Selling_Price:", (df['Selling_Price'] < 0).sum())
print()
    #Check for unrealistic Year values
print(df[(df['Year'] < 1980) | (df['Year'] > 2026)][['Make', 'Model', 'Year']])
print()
    #Check cars not electric should have Fuel_Size = 0
print(df[(df['Engine_Size'] == 0) & (df['Fuel_Type'] != 'Electric')])
```

```python
#step13 data type conversion
print(df.dtypes)

df['Accident_History'] = df['Accident_History'].astype(int)

categorical_cols = ['Make', 'Model', 'Fuel_Type', 'Transmission', 'Service_History',
                     'Color', 'Body_Type', 'Drivetrain', 'Location']
for col in categorical_cols:
    df[col] = df[col].astype('category')

print(df.dtypes)
```

Part 4: Feature Engineering

```python
#view unique values for categorical columns
categorical_cols = ['Make', 'Model', 'Fuel_Type', 'Transmission', 'Service_History',
                     'Color', 'Body_Type', 'Drivetrain', 'Location']

for col in categorical_cols:
    print(f"{col}: {df[col].nunique()} unique values")
```

```python
#step14 label encoding
# Model, Transmission
from sklearn.preprocessing import LabelEncoder

le_model = LabelEncoder()
df['Model_enc'] = le_model.fit_transform(df['Model'])

le_transmission = LabelEncoder()
df['Transmission_enc'] = le_transmission.fit_transform(df['Transmission'])

print(df[['Transmission','Transmission_enc']].drop_duplicates())
print(df[['Model','Model_enc']].drop_duplicates().sort_values('Model_enc'))
print("Unique Model Encoded Values:", df['Model_enc'].nunique())
print("Unique Transmission Encoded Values:", df['Transmission_enc'].nunique())
```

```python
#step15 one-hot encodeing
# Make, Fuel_Type, Color, Body_Type, Drivetrain, Location
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder(sparse_output=False)
onehot_cols = ['Make', 'Fuel_Type', 'Color', 'Body_Type', 'Drivetrain', 'Location']

enc_arr = ohe.fit_transform(df[onehot_cols])
print("New feature names:", ohe.get_feature_names_out(onehot_cols))

# Convert the encoded array into a DataFrame with proper column names
df_enc = pd.DataFrame(enc_arr, columns=ohe.get_feature_names_out(onehot_cols), index=df.index)

# Combine with the main dataframe
df = pd.concat([df, df_enc], axis=1)

print(df.head())
```

```python
#step16 ordinal encodeing
# Service_History
# No Service < Partial Service < Full Service
from sklearn.preprocessing import OrdinalEncoder
enc = OrdinalEncoder(categories=[
    ['No Service', 'Partial Service', 'Full Service']
])

df['Service_History_enc'] = enc.fit_transform(df[['Service_History']])

print(df[['Service_History','Service_History_enc']].drop_duplicates())
print("Unique Service_History Encoded Values:", df['Service_History_enc'].nunique())

print("Total columns now:", df.shape[1])
print(df.columns.tolist())
```

Automobile Data after Feature Engineering.

```python
# Select only numeric columns (encoded + original numeric ones)
numeric_and_encoded = df.select_dtypes(include=['number']).columns.tolist()

# Calculate correlation with Selling_Price only, sorted from strongest to weakest
corr_with_price = df[numeric_and_encoded].corr()['Selling_Price'].sort_values(ascending=False)

plt.figure(figsize=(8, 12))
corr_with_price.drop('Selling_Price').plot(kind='barh')
plt.title('Correlation with Selling Price')
plt.xlabel('Correlation Coefficient')
plt.axvline(x=0, color='black', linewidth=0.8)
plt.tight_layout()
plt.show()
```

```python
# Select numeric + encoded columns that make sense to visualize together
# (excluding one-hot columns since there are too many to fit in a readable heatmap)
selected_cols = ['Year', 'Engine_Size', 'Mileage', 'Horsepower', 'Torque', 'Owners',
                  'Accident_History', 'Fuel_Efficiency', 'Model_enc',
                  'Transmission_enc', 'Service_History_enc', 'Selling_Price']

corr = df[selected_cols].corr()

plt.figure(figsize=(11, 9))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1)
plt.title('Correlation Heatmap (Numeric + Encoded Features)')
plt.tight_layout()
plt.show()
```
