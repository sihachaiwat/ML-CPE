```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

#Part1 dataset exploration
#step1 load dataset
df = pd.read_csv("automobile_dataset.csv")
print(df.head())
```

```python
#step2 display shape
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
```

```python
#step3 display datatypes
print(df.info())
```

```python
import warnings
warnings.filterwarnings('ignore')
#step4 display summary statistics
print(df.describe())
print(df.describe(include='object'))
```

```python
#step5 display missing values
print(df.isnull().sum())
```

```python
#step6 display duplicate records
print('duplicated: ',df.duplicated().sum())
print(df[df.duplicated()])
```

```python
#step7 displaay class ditribution
# Distribution of Fuel_Type
print(df['Fuel_Type'].value_counts(normalize=True)*100)
print()
print(df['Make'].value_counts(normalize=True)*100)
print()
print(df['Transmission'].value_counts(normalize=True)*100)
print()
print(df['Body_Type'].value_counts(normalize=True)*100)
print()
```

```python
#Part2 data visualization
#step8 histogram
# Histogram of numeric columns
numeric_cols = ['Year', 'Engine_Size', 'Mileage', 'Horsepower', 'Torque', 'Owners', 'Fuel_Efficiency', 'Selling_Price']

df[numeric_cols].hist(figsize=(14, 10), bins=15, edgecolor='black')
plt.tight_layout()
plt.show()

plt.figure(figsize=(6, 4))
sns.histplot(df['Fuel_Efficiency'], bins=20, kde=True)
plt.title('Distribution of Fuel Efficiency')
plt.show()

plt.figure(figsize=(6, 4))
sns.histplot(df['Selling_Price'], bins=20, kde=True)
plt.title('Distribution of Selling Price')
plt.show()
```

```python
# Check rows with unusually high Fuel_Efficiency
outliers = df[df['Fuel_Efficiency'] > 60]
print("Number of outlier rows:", len(outliers))
print(outliers['Fuel_Type'].value_counts())
print(outliers[['Make', 'Model', 'Fuel_Type', 'Fuel_Efficiency']].head(10))
```

```python
#step9 correlation heatmap
numeric_cols = ['Year', 'Engine_Size', 'Mileage', 'Horsepower', 'Torque', 'Owners', 'Accident_History', 'Fuel_Efficiency', 'Selling_Price']

# Calculate correlation matrix
corr = df[numeric_cols].corr()
print(corr)

# Plot heatmap
plt.figure(figsize=(9, 7))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

```python
#step10 missing value handling
# fill with median
numeric_missing = ['Engine_Size', 'Horsepower', 'Torque', 'Accident_History', 'Fuel_Efficiency']
for col in numeric_missing:
    df[col] = df[col].fillna(df[col].median())

# fill with mode 
categorical_missing = ['Transmission', 'Service_History', 'Color', 'Location']
for col in categorical_missing:
    df[col] = df[col].fillna(df[col].mode()[0])
    
print(df.isnull().sum())
```

```python
#step11 duplicate removal
print('Num of duplicated',df.duplicated().sum())

#dont have any duplicates so no need to remove
```

```python
#step12 incorrect data correction
    #Check for negative values
print("Negative Year:", (df['Year'] < 0).sum())
print("Negative Mileage:", (df['Mileage'] < 0).sum())
print("Negative Engine_Size:", (df['Engine_Size'] < 0).sum())
print("Negative Horsepower:", (df['Horsepower'] < 0).sum())
print("Negative Selling_Price:", (df['Selling_Price'] < 0).sum())
print()
    #Check for unrealistic Year values
print(df[(df['Year'] < 1980) | (df['Year'] > 2026)][['Make', 'Model', 'Year']])
print()
    #Check cars not electric should have Fuel_Size = 0
print(df[(df['Engine_Size'] == 0) & (df['Fuel_Type'] != 'Electric')])
```

```python
#step13 data type conversion
print(df.dtypes)

df['Accident_History'] = df['Accident_History'].astype(int)

categorical_cols = ['Make', 'Model', 'Fuel_Type', 'Transmission', 'Service_History',
                     'Color', 'Body_Type', 'Drivetrain', 'Location']
for col in categorical_cols:
    df[col] = df[col].astype('category')

print(df.dtypes)
```

```python
#Part4 feature engineering
#step14 lebel encoding
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df['Gender_encoded'] = le.fit_transform(df['Gender'])

print("Gender mapping:", dict(zip(le.classes_, le.transform(le.classes_))))

le_breed = LabelEncoder()
df['Breed_encoded'] = le_breed.fit_transform(df['Breed'])

print("Breed mapping:")
print(dict(list(zip(le_breed.classes_, le_breed.transform(le_breed.classes_)))[:5]))

df[['Gender', 'Gender_encoded', 'Breed', 'Breed_encoded']].head()
```

```python
#step15 one hot encodeing
color_encoded = pd.get_dummies(df['Color'], prefix='Color')

print("Num of new columns:", color_encoded.shape[1])
color_encoded.head()

#combine the one-hot encoded columns with the original df
df = pd.concat([df, color_encoded], axis=1)

print(df.head())
```
